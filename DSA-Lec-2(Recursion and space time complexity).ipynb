# 🚀 Ultimate Time Complexity Master Sheet: 


## 1. Core Mathematical Definitions & Exact Boundaries

* **$O(n)$ (Big O - Asymptotic Upper Bound):** $f(n) \le c \cdot g(n)$ for all $n \ge n_0$. Guarantees the algorithm will *never* perform worse than this curve.
* **$\Omega(n)$ (Big Omega - Asymptotic Lower Bound):** $f(n) \ge c \cdot g(n)$ for all $n \ge n_0$. Guarantees the absolute *minimum* work required.
* **$\Theta(n)$ (Big Theta - Asymptotic Tight Bound):** $c_1 \cdot g(n) \le f(n) \le c_2 \cdot g(n)$. The upper and lower bounds match; behavior is tightly locked.
* **$o(n)$ (Little o - Strict Upper Bound):** $f(n) < c \cdot g(n)$. Dominated entirely by $g(n)$ asymptotically (e.g., $n = o(n^2)$).
* **$\omega(n)$ (Little Omega - Strict Lower Bound):** $f(n) > c \cdot g(n)$. Strictly grows faster than $g(n)$.

## 3. The 4 Input-Dependent Sub-Cases
An algorithm's performance can swing wildly based on structural differences in inputs of identical sizes.

*   **Best Case:** Ideal structural state requiring minimum possible operations.
*   **Average Case:** Expected behavior modeled over uniform distribution of all potential inputs.
*   **Worst Case:** Absolute bottleneck arrangement producing maximum executions.
*   **Amortized Case:** Average operational cost over a sequential stream of executions where a high-cost execution is mathematically balanced out by guaranteed low-cost steps.

### Code Demonstration: Comprehensive Sub-Cases
```python
def linear_search_and_insert(dynamic_list, target):
    # Part 1: Search Case Analysis
    for index, item in enumerate(dynamic_list):
        if item == target:
            return index # Best: O(1) [First item] | Worst: O(n) [Last item/Absent]
            
    # Part 2: Structural Trapping Case (Amortized Analysis)
    # Assume dynamic_list triggers an internal copy block when max capacity hit
    dynamic_list.append(target) 
    return len(dynamic_list) - 1

## 4. Advanced Scenarios & Multi-Variable Dimensions

### 💡 Expected Amortized vs. Pure Amortized
*   **Pure Amortized (Dynamic Array):** Guaranteed mathematically. Array doubling at \(n\) capacity costs \(O(n)\), but the preceding \(n-1\) allocations cost \(O(1)\). Total sequence cost \(2n \implies\) Strictly \(O(1)\) per operation.
*   **Expected Amortized (Hash Table):** Relies on probability. If a Hash Map encounters massive **Hash Collisions** (multiple keys hitting the same memory bucket), the structure degrades into a Linked List or Balanced Tree. Insertion can degrade to \(O(n)\) worst-case. With a good hash function, the *expected amortized* cost is \(O(1)\).

# 🗺️ Graphs, Matrices & Multi-Variables

Never default variables to ($n$). Explicitly map inputs to structural parameters.

* **Graph Traversals (BFS / DFS):** $O(V + E)$ where $V = \text{Vertices}$ and $E = \text{Edges}$. If dense ($E \approx V^2$), complexity transforms to $O(V^2)$.
* **Grid Search (A* / Dijkstra on Grids):** $O(R \times C \log(R \times C))$ where $R = \text{Rows}$ and $C = \text{Columns}$.
* **String Matching (KMP Algorithm):** $O(N + M)$ where $N$ is text length and $M$ is pattern length.

# 🌲 Tree Recurrences & Master Theorem Mechanics

For Divide-and-Conquer algorithms obeying $T(n) = aT(n/b) + f(n)$:

* **Case 1 ($f(n)$ dominated):** $f(n) = O(n^{\log_b a - \epsilon}) \implies T(n) = \Theta(n^{\log_b a})$
* **Case 2 (Balanced):** $f(n) = \Theta(n^{\log_b a}) \implies T(n) = \Theta(n^{\log_b a} \log n)$
* **Case 3 ($f(n)$ dominating):** $f(n) = \Omega(n^{\log_b a + \epsilon}) \implies T(n) = \Theta(f(n))$
* **Exception (Binary Search):** $T(n) = T(n/2) + O(1) \implies a=1, b=2 \implies O(\log n)$
* **Exception (Merge Sort):** $T(n) = 2T(n/2) + O(n) \implies a=2, b=2 \implies O(n \log n)$

## 5. Hidden Engine Costs & Underlying System Realities

### 🐍 Built-In Language Shifting Traps (Python Core Engine)
High-level code reads as simple commands but runs deep loops underneath:
*   **Deletion/Insertion Shifts:** `list.pop(0)` or `list.insert(0, val)` forces the underlying memory engine to shift every trailing reference block in contiguous memory \(\implies O(n)\).
*   **String Immutability Copying:** `s += "char"` cannot modify memory in-place. It allocates a brand-new string container of size \(len(s) + 1\) and copies old characters \(\implies O(n)\) per iteration. A loop doing this runs at \(O(n^2)\).
*   **Timsort (Python `sort()`):** Uses \(O(n)\) in best-case (already sorted data via natural runs), but hits \(O(n \log n)\) worst-case.

### 🖥️ Hardware Microarchitectures & Cache Realities
*   **Cache Locality:** An \(O(n)\) iteration through a flat, contiguous memory block (Array) loads entire lines into the CPU cache at once. A Linked List with matching \(O(n)\) operations requires jumping random memory pointers, forcing continuous CPU **Cache Misses**. The array will run up to 100x faster in real production execution.
*   **The Constant Factor Mirage:** Algorithm A does \(10^7 \cdot n\) steps. Algorithm B does \(5 \cdot n^2\) steps. Asymptotically, A is superior (\(O(n) < O(n^2)\)). Practically, Algorithm B executes faster for all inputs where \(n < 2,000,000\).

# Real Complexity: O(n + m) -> Boundaries are completely independent!

    print(item)
```

### ❌ Trap 2: Loop Variable Growth Mutations
```python

# Real Complexity: O(log n) -> The iterator grows exponentially!

while i < n:
    print(i)
    i *= 2 
```

### ❌ Trap 3: Sub-Slicing Allocations in Recursive Trees
```python
# False Alarm: Looks like O(log n) because it is cutting arrays in half.
# Real Complexity: O(n) -> Memory copying overrides performance gains!
def find_midpoint_val(arr):
    if not arr: return None
    mid = len(arr) // 2
    # arr[:mid] creates a brand-new array copy of size (n/2) across memory!
    return arr[mid] + find_midpoint_val(arr[:mid]) 
```

### ❌ Trap 4: Tricked by Logarithmic Pointers
```python

# The inner loop pointer 'j' is never reset to 0! It only moves forward a total of 'n' times across the lifetime of the program.
j = 0

    while j < n and array[j] < boundary_condition:
        j += 1
```

# Types of Time complexity:
# ⏱️ Understanding Time Complexity & Big O Notation

Time complexity measures **how the runtime of an algorithm scales as the input size ($N$) grows**. Instead of measuring time in seconds (which changes based on the computer hardware), we count the **number of operational steps** the code takes.

---

## 🛑 Why Big O Ignores Constant Terms
Big O notation is entirely focused on the **growth rate (the trend line)** of an algorithm as data approaches infinity ($N \to \infty$). 
* A loop running $N$ times and a loop running $2N + 5$ times both grow at the exact same **linear rate**.
* If the input size multiplies by 10, the runtime multiplies by 10 for both cases. 
* Constants shift the line up or down on a graph, but they **do not change its fundamental shape**. Therefore, constants are dropped.

---

## 🔍 The Big O Cheat Sheet

| Time Complexity | Big O | How It Scales | Python Code Sign | Real-World Analogy |
| :--- | :--- | :--- | :--- | :--- |
| **Constant** | $O(1)$ | Stays exactly the same, no matter how much data you add. | No loops; accessing a list index or dictionary key. | Checking *only* the first egg in a carton. |
| **Linear** | $O(N)$ | Grows at the exact same rate as the data. (10x data = 10x time). | A single `for` or `while` loop running up to $N$. | Checking *every single egg* in a carton once. |
| **Quadratic** | $O(N^2)$ | Explodes rapidly. (10x data = 100x time). | Nested loops (a loop inside another loop) over the same data. | Comparing *every egg with every other egg* to find the heaviest. |

---

## 🛠️ How to Analyze Code in 3 Steps

1. **Identify the Input ($N$):** Find the variable or data collection that can grow infinitely large (like the length of a list).
2. **Count the Loops:** 
   * No loops = $O(1)$
   * One loop = $O(N)$
   * Nested loops = Multiply them ($O(N \times N) = O(N^2)$)
3. **Drop the Garbage:** Keep only the fastest-growing term and delete all constants or smaller terms (e.g., $O(3N^2 + 5N + 12)$ simplifies directly to **$O(N^2)$**).
---
### 🔍 Core Check: Does this code loop through my data?

* ❌ **No loops** $\rightarrow$ **Constant Time** $O(1)$
* 🏃 **One loop** $\rightarrow$ **Linear Time** $O(N)$
* 🌀 **Loop inside a loop (Nested)** $\rightarrow$ **Quadratic Time** $O(N^2)$

> 💡 **Important Catch (Hidden Loops):** Watch out for built-in Python functions! Even if you don't see a `for` or `while` keyword, operations like `item in my_list`, `max(my_list)`, or `my_list.count(x)` loop through data under the hood. They are secretly $O(N)$ operations!

## 🪵 Logarithmic Time Complexity \(O(\log N)\)

Logarithmic complexity represents algorithms that scale incredibly efficiently. As the input data grows exponentially, the execution time only grows linearly. 

---

### 🧠 The Core Logic

An algorithm is \(O(\log N)\) if it **reduces the size of the remaining problem by a constant fraction (usually half)** at every single step. 

* **The Math:** If \(N = 8\), the problem splits: \(8 \to 4 \to 2 \to 1\) (3 steps). Mathematically, \(\log_2(8) = 3\).
* **The Python Code Sign:** While a linear loop \(O(N)\) increments step-by-step (`i += 1`), a logarithmic loop changes its search boundaries or data size by **multiplication or division** (`high = mid - 1` or `n //= 2`).

---

### 📋 All Structural Cases & Variations

#### Case 1: Pure Logarithmic — \(O(\log N)\)
The search space splits perfectly in half each time.
* **Example:** Searching for a specific number in a sorted array using Binary Search.
* **Performance:** Takes roughly 20 steps for 1 million items.

#### Case 2: Log-Linear (Linearithmic) — \(O(N \log N)\)
An \(O(\log N)\) operation is performed \(N\) times. 
* **Example:** Efficient sorting algorithms like Merge Sort, Quick Sort, or Python’s built-in `.sort()` (Timsort). 
* **Performance:** The gold standard baseline for sorting data structurally.

#### Case 3: Base Changes — \(O(\log_3 N)\) or \(O(\log_k N)\)
The problem is divided into 3 parts (or \(k\) parts) instead of 2 parts at each step.
* **Important Rule:** In Big O notation, we **drop the base constants**. \(O(\log_3 N)\) simplifies directly to **\(O(\log N)\)** because \(\log_a N = \frac{\log_b N}{\log_b a}\) (the base change results in a constant multiplier, which is ignored).

---

### 🐍 Practical Python Examples

#### Example 1: Binary Search (\(O(\log N)\))
Finding a value inside an already **sorted** list by continually cutting the list in half.

```python
def binary_search(sorted_list, target):
    low = 0
    high = len(sorted_list) - 1

    while low <= high:
        mid = (low + high) // 2  # Find the middle element
        guess = sorted_list[mid]

        if guess == target:
            return mid       # Target found
        elif guess > target:
            high = mid - 1   # Cut the upper half out
        else:
            low = mid + 1    # Cut the lower half out
    return -1
```

#### Example 2: Repeated Division (\(O(\log N)\))
A numeric process where a number is broken down until it hits a base limit.

```python
def count_divisions(n):
    steps = 0
    while n > 1:
        n //= 2              # Divides the problem space in half
        steps += 1
    return steps
```

# 📈 Advanced Time Complexity Guide: From Optimal to Worst-Case

This section details all remaining major time complexities, ranging from fast log-linear sorting to catastrophic exponential and factorial scaling.

---

## 🏎️ 1. Log-Linear (Linearithmic) Time — \(O(N \log N)\)

### 🧠 The Core Logic
This occurs when an algorithm splits a problem into logarithmic chunks (\(O(\log N)\)), but has to look through the entire dataset (\(O(N)\)) at each split level. Think of it as running a logarithmic operation \(N\) times.

* **The Python Sign:** Divide-and-conquer algorithms, or a linear loop containing a nested logarithmic binary search inside it.
* **The Trick:** Whenever you see **sorting** or **efficient tree rebuilding**, think \(O(N \log N)\). 

### 🐍 Python Example: Merge Sort Baseline
```python
def merge_sort(my_list):
    if len(my_list) <= 1:
        return my_list
    
    # Logarithmic Splitting: O(log N) depth
    mid = len(my_list) // 2
    left = merge_sort(my_list[:mid])
    right = merge_sort(my_list[mid:])
    
    # Linear Merging: O(N) work per level
    return merge(left, right) 
```
* **Built-in Alert:** Python’s native `my_list.sort()` and `sorted(my_list)` use an algorithm called Timsort, which runs at **\(O(N \log N)\)**.

---

## 🌋 2. Cubic Time — \(O(N^3)\)

### 🧠 The Core Logic
The execution time grows relative to the cube of the input size. If your data grows \(10\times\), your runtime explodes by \(1,000\times\) (\(10 \times 10 \times 10\)).

* **The Python Sign:** Three deeply nested loops traversing the same dataset.
* **The Trick:** Often found in basic 3D graphics rendering matrix loops or naive physics engine calculations checking triplets of items.

### 🐍 Python Example: Naive Triplet Finder
```python
def find_triplet_sum(my_list, target):
    # Three nested loops looking at the same data array
    for i in my_list:
        for j in my_list:
            for k in my_list:
                if i + j + k == target:
                    return (i, j, k)
    return None
```

---

## 💣 3. Exponential Time — \(O(2^N)\)

### 🧠 The Core Logic
The runtime **doubles** with every single addition to the input size. If \(N=10\), it takes around 1,000 operations. If \(N=20\), it takes over 1 million operations! This complexity becomes completely unusable for inputs greater than 40.

* **The Python Sign:** Recursive functions that make **two recursive calls** at each step.
* **The Trick:** Look for combinations, brute-force password cracking, or solving subsets of combinations.

### 🐍 Python Example: Naive Fibonacci Code
```python
def naive_fibonacci(n):
    if n <= 1:
        return n
    # Explodes because it spawns two new function instances per call
    return naive_fibonacci(n - 1) + naive_fibonacci(n - 2)
```
* **Optimization Trick:** You can fix \(O(2^N)\) recursion bottlenecks by utilizing **Memoization** (storing old results using Python's `@lru_cache` decorator), which magically drops the complexity down to a flat linear **\(O(N)\)**.

---

## 🚨 4. Factorial Time — \(O(N!)\)

### 🧠 The Core Logic
The absolute worst runtime footprint. The execution time multiplies by every integer up to \(N\) (\(N \times (N-1) \times (N-2) \dots\)). An input of just \(N=20\) would take your modern computer thousands of years to calculate.

* **The Python Sign:** Generating every single possible arrangement, permutation, or routing ordering.
* **The Trick:** The classic indicator is the **Traveling Salesperson Problem (TSP)** or any function that explicitly uses `itertools.permutations`.

### 🐍 Python Example: Brute-Force Permutations
```python
import itertools

def break_traveling_salesman(cities):
    # Generates N! paths to evaluate every possible route order
    all_possible_routes = itertools.permutations(cities)
    
    best_route = None
    for route in all_possible_routes:
        check_distance(route) # Processes N! iterations
```

## 🗂️ Complete Big O Hierarchy Cheat Sheet

Here is the master hierarchy from **best performance to catastrophic performance**:

$$O(1) < O(\log N) < O(N) < O(N \log N) < O(N^2) < O(N^3) < O(2^N) < O(N!)$$

| Complexity | Operations ($N=10$) | Operations ($N=100$) | Code Diagnostic | Status |
| :--- | :--- | :--- | :--- | :--- |
| **$O(1)$** | 1 | 1 | Direct Indexing / Basic Math | 🟢 Excellent |
| **$O(\log N)$** | ~3 | ~7 | Divided Workspace (`// 2`) | 🟢 Excellent |
| **$O(N)$** | 10 | 100 | Single Loop (`for item in data`) | 🟡 Good |
| **$O(N \log N)$** | ~33 | ~664 | Sorting / Divide & Conquer | 🟡 Acceptable |
| **$O(N^2)$** | 100 | 10,000 | Double Nested Loop | 🔴 Slow |
| **$O(2^N)$** | 1,024 | $1.26 \times 10^{30}$ | Multiple Branching Recursion | ❌ Horrible |
| **$O(N!)$** | 3,628,800 | Infinity (Practically) | Permutations & Combinations | ❌ Catastrophic |

# 🧠 Complete Guide to Space Complexity

Space complexity measures **how much extra memory (RAM) an algorithm needs to run as the input size (\(N\)) grows**. Just like time complexity, we use Big O notation to describe this scaling behavior.

---

## ⚖️ Total Space vs. Auxiliary Space

When analyzing space complexity, there is a technical distinction you must know:
1. **Total Space Complexity:** Input Space + Auxiliary Space.
2. **Auxiliary Space (Extra Space):** The temporary memory allocated by the algorithm *while* it runs.

> 💡 **Interview Trick:** When interviewers ask for "space complexity," they almost always mean **Auxiliary Space**. They want to know how much *extra* memory your specific logic creates.

---

## 🔍 The Space Complexity Hierarchy & Cases

### 🛑 1. Constant Space — \(O(1)\)
The memory footprint stays exactly the same, no matter how large the input grows. You are reusing the same variables over and over.

* **The Python Sign:** Fixed variables, modifying data "in-place" without cloning it, or basic mathematical calculations.
* **The Trick:** If you can solve a problem by overwriting elements in the original list instead of making a new list, you achieve \(O(1)\) space.

```python
def find_max_and_min(my_list):
    # This creates exactly two variables, regardless of list size.
    max_val = my_list[0]               # O(1) space
    min_val = my_list[0]               # O(1) space
    
    for num in my_list:
        if num > max_val: max_val = num
        if num < min_val: min_val = num
        
    return max_val, min_val            # Auxiliary Space: O(1)
```

---

### 🏃 2. Linear Space — \(O(N)\)
The extra memory scales directly with the input size. If your input doubles, your memory usage doubles.

* **The Python Sign:** Creating a new list, dictionary, set, or string that stores a copy or modification of your input items.
* **The Call Stack Sign:** Normal recursion that goes \(N\) layers deep.

```python
def double_the_list(my_list):
    new_list = []                      # Spawns a new container
    for item in my_list:
        new_list.append(item * 2)      # Will hold exactly N elements
    return new_list                    # Auxiliary Space: O(N)
```

---

### 🌀 3. Quadratic Space — \(O(N^2)\)
The memory footprint explodes relative to the square of the input. 

* **The Python Sign:** Creating a 2D grid/matrix of size \(N \times N\), or creating nested collections where each item contains a copy of the list.

```python
def create_multiplication_table(n):
    # Generates a matrix containing N rows and N columns
    matrix = [[i * j for j in range(n)] for i in range(n)]
    return matrix                      # Auxiliary Space: O(N^2)
```

---

## 🪤 Dangerous Traps & Hidden Space Behaviors

Python handles memory dynamically, which sets up several major traps for beginners.

### 🚨 Trap 1: The Invisible Call Stack (Recursion Trap)
Even if your recursive function does not declare a single list or variable, **it still consumes memory**. Every time a function calls itself, Python pushes a "stack frame" into memory to remember where it left off.

```python
def countdown(n):
    if n <= 0:
        return
    countdown(n - 1)  # Hidden Trap! 
```
* **Analysis:** If \(N = 10,000\), Python must store 10,000 incomplete function execution details in RAM. 
* **Space Complexity:** **\(O(N)\)**, even though there are no arrays!

### 🚨 Trap 2: String Concatenation Inside Loops
Strings in Python are **immutable** (they cannot be changed in place). Every time you modify a string, Python silently copies the old string over to a brand new memory location.

```python
def naive_join(my_list):
    result = ""
    for word in my_list:
        result += word  # Hidden Trap! Generates a completely new string every loop
    return result
```
* **Space Complexity:** **\(O(N^2)\)** in terms of intermediate memory churn!
* **The Trick/Fix:** Use `''.join(my_list)`. It calculates the final required space upfront, ensuring **\(O(N)\)** space.

### 🚨 Trap 3: List Slicing (`my_list[start:end]`)
Slicing a list in Python does not just point to the old data; it **duplicates** that slice into a brand new memory array.

```python
def process_data(my_list):
    # If you pass a sliced list into a recursive step, you clone data
    left_half = my_list[:len(my_list)//2]  # Duplicates half the elements into RAM
```
* **The Trick/Fix:** Instead of slicing data arrays, pass tracking **indices** (`low` and `high` pointer integers) to your function. Pointers consume $O(1)$ space.

---

## 🛠️ The Ultimate Space Optimization Blueprint

Want to lower your space complexity in Python? Memorize these rules:

1. **Use Generators (`yield`):** If you are processing a massive file, don't read it all into a list. Use `yield` to stream data one item at a time. This drops space usage from $O(N)$ to $O(1)$.
2. **Mutate In-Place:** If you don't need the original raw data anymore, overwrite it directly (`my_list[i] = new_value`) instead of appending to a fresh list.
3. **Prefer Loops Over Recursion:** If an algorithm can be written easily with a `while` or `for` loop, choose it. Loops don't create call stack memory blocks.
